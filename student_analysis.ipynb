```python
import pandas as pd
# -------------------------------
# 1. FILE HANDLING WITH EXCEPTION
# -------------------------------
def load_data(file_path):
    try:
        df = pd.read_csv(file_path, sep=';', encoding='latin1')
        df.columns = df.columns.str.strip()
        print("File loaded successfully!\n")
        print("Columns found:")
        print(df.columns.tolist())
        required_cols = ['G1 ', 'G2', 'G3', 'studytime', 'absences']
        for col in required_cols:
            if col not in df.columns:
                print(f"ERROR: Column '{col}' not found in file.")
                return None
        return df
    except Exception as e:
        print("Error while reading file:", e)
        return None
# -------------------------------
# 2. DATA EXPLORATION
# -------------------------------
def explore_data(df):
    print("\nFirst 10 rows:\n")
    print(df.head(10))
    print("\nLast 5 rows:\n")
    print(df.tail(5))
    print("\nDataFrame Info:\n")
    print(df.info())
    print("\nStudents with G3 < 10 OR absences > 10:\n")
    filtered = df[(df['G3'] < 10) | (df['absences'] > 10)]
    print(filtered)
# -------------------------------
# 3. HANDLE MISSING VALUES
# -------------------------------
def handle_missing_values(df):
    columns = ['G3', 'studytime', 'absences']
    for col in columns:
        df[col] = df[col].fillna(df[col].mean())
    print("\nMissing values handled.\n")
# -------------------------------
# 4. CREATE NEW COLUMNS
# -------------------------------
def create_new_columns(df):
    df['StudyEfficiency'] = df.apply(
        lambda row: row['G3'] / row['studytime'] if row['studytime'] != 0 else 0,
        axis=1
    )
    df['GradeImprovement'] = df['G3'] - df['G1']
    def family_support(row):
        if row['famsup'] == 'yes' and (row['Medu'] >= 3 or row['Fedu'] >= 3):
            return "High"
        elif row['famsup'] == 'yes':
            return "Medium"
        else:
            return "Low"
    df['FamilySupportScore'] = df.apply(family_support, axis=1)
    print("\nNew columns added.\n")
# -------------------------------
# 5. PARENTAL EDUCATION LEVEL
# -------------------------------
def parental_education_analysis(df):
    def categorize(row):
        avg = (row['Medu'] + row['Fedu']) / 2
        if avg <= 2:
            return "Low"
        elif avg <= 3:
            return "Medium"
        else:
            return "High"
    df['EducationLevel'] = df.apply(categorize, axis=1)
    result = df.groupby('EducationLevel')['G3'].mean()
    print("\nAverage G3 by Education Level:")
    print(result)
# -------------------------------
# 6. GRADE DISTRIBUTION
# -------------------------------
def grade_distribution(df):
    fail = len(df[df['G3'] < 10])
    pass_ = len(df[(df['G3'] >= 10) & (df['G3'] <= 15)])
    good = len(df[df['G3'] > 15])
    print("\nGrade Distribution:")
    print("Fail:", fail)
    print("Pass:", pass_)
    print("Good:", good)
# -------------------------------
# 7. GROUP BY SCHOOL
# -------------------------------
def group_by_school(df):
    gp = df[df['school'] == 'GP']
    ms = df[df['school'] == 'MS']
    gp_eff = (gp['G3'] / gp['studytime']).mean()
    ms_eff = (ms['G3'] / ms['studytime']).mean()
    print("\nGP Avg Efficiency:", gp_eff)
    print("MS Avg Efficiency:", ms_eff)
# -------------------------------
# 8. STATISTICS ANALYSIS
# -------------------------------
def statistics_analysis(df):
    columns = ['studytime', 'absences', 'G1', 'G2', 'G3']
    for col in columns:
        print(f"\nStatistics for {col}:")
        print("Mean:", df[col].mean())
        print("Median:", df[col].median())
        print("Mode:", df[col].mode()[0])
        print("Min:", df[col].min())
        print("Max:", df[col].max())
    print("\nOverall Statistics:\n")
    print(df.describe())
# -------------------------------
# 9. GENERATE REPORT
# -------------------------------
def generate_report(df):
    try:
        with open("student_performance_analysis_report.txt", "w") as f:
            f.write("STUDENT PERFORMANCE ANALYSIS REPORT\n")
            f.write("=" * 40 + "\n")
            f.write(f"\nAverage Final Grade (G3): {df['G3'].mean()}")
            f.write(f"\nMax G3: {df['G3'].max()}")
            f.write(f"\nMin G3: {df['G3'].min()}")
            fail = len(df[df['G3'] < 10])
            f.write("\n\nGrade Distribution:\n")
            f.write(f"Fail: {fail}")

        print("\nReport generated successfully!")
        print("Check your folder for 'student_performance_analysis_report.txt'")

    except Exception as e:
        print("Error writing report:", e)
# -------------------------------
# MAIN FUNCTION
# -------------------------------
def main():
    file_path = "student-mat.csv"
    df = load_data(file_path)
    if df is not None:
        explore_data(df)
        handle_missing_values(df)
        create_new_columns(df)
        parental_education_analysis(df)
        grade_distribution(df)
        group_by_school(df)
        statistics_analysis(df)
        generate_report(df)
if __name__ == "__main__":
    main()
```
